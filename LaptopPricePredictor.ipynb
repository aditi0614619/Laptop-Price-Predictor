```python
# a function which helps us to break the Memory into a few columns HDD SSD Hybrid Flash_Storage
def get(text):
    text = str(text)
    num = ""
    for ch in text:
        if ch.isdigit() or ch == '.':
            num += ch
        else:
            break

    if num == "":
        return 0

    val = float(num)

    if "TB" in text:
        return int(val * 1024)
    else:  # GB
        return int(val)
def ssd(text):
    lst=text.split()
    ans=0
    for i in range(len(lst)):
        if(lst[i]=="SSD"):
            ans+=get(lst[i-1])
    return ans
def hdd(text):
    ans=0;
    lst=text.split()
    for i in range(len(lst)):
        if(lst[i]=="HDD"):
            ans+=get(lst[i-1])
    return ans
def hybrid(text):
    lst=text.split()
    ans=0
    for i in range(len(lst)):
        if(lst[i]=="Hybrid"):
            ans+=get(lst[i-1])
    return ans
def Flash(text):
    lst=text.split()
    ans=0
    for i in range(len(lst)):
        if(lst[i]=="Flash"):
            ans+=get(lst[i-1])
    return ans
```

```python
df['SSD']=df['Memory'].apply(ssd)
df['HDD']=df['Memory'].apply(hdd)
df['hybrid']=df['Memory'].apply(hybrid)
df['Flash_Storage']=df['Memory'].apply(Flash)
df.head()
```
